## Завдання 3. Скалярний добуток і норма вектора

Для векторів `u = [3, -1, 2]` та `v = [1, 4, -2]`:

1. Обчисліть скалярний добуток.
2. Знайдіть евклідову норму кожного вектора.
3. Обчисліть косинус кута між векторами за формулою

\[
\cos(\alpha)=\frac{u\cdot v}{\|u\|\|v\|}
\]


````python
u = np.array([3, -1, 2])
v = np.array([1, 4, -2])

dot_product = np.dot(u, v)

norm_u = np.linalg.norm(u)
norm_v = np.linalg.norm(v)

cos_alpha = dot_product / (norm_u * norm_v)

print("Скалярний добуток (u · v):", dot_product)
print("Норма вектора u:", norm_u)
print("Норма вектора v:", norm_v)
print("Косинус кута cos(α):", cos_alpha)

````

## Завдання 7. Операції з матрицями

Нехай задано дві матриці:

\[
A =
\begin{bmatrix}
1 & 2\\
3 & 4
\end{bmatrix},
\quad
B =
\begin{bmatrix}
5 & 6\\
7 & 8
\end{bmatrix}
\]

Обчисліть:
- `A + B`;
- `A - B`;
- поелементний добуток;
- матричний добуток `A @ B`;
- транспоновану матрицю `A`.

````python
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

sum_AB = A + B

diff_AB = A - B

elem_prod = A * B

matmul_AB = A @ B

trans_A = A.T

print("Матриця A:\n", A)
print("\nМатриця B:\n", B)
print("\n1. Сума A + B:\n", sum_AB)
print("\n2. Різниця A - B:\n", diff_AB)
print("\n3. Поелементний добуток (A * B):\n", elem_prod)
print("\n4. Матричний добуток (A @ B):\n", matmul_AB)
print("\n5. Транспонована матриця A (A^T):\n", trans_A)
````

## Завдання 8. Визначник та обернена матриця

Для матриці

\[
A =
\begin{bmatrix}
4 & 7\\
2 & 6
\end{bmatrix}
\]

1. Знайдіть визначник.
2. Знайдіть обернену матрицю.
3. Перевірте результат множенням `A @ A_inv`.
4. Поясніть у коментарі, який результат має бути отриманий.

````python
A = np.array([[4, 7],
              [2, 6]])

det_A = np.linalg.det(A)

A_inv = np.linalg.inv(A)

check_product = A @ A_inv

print("Матриця A:\n", A)
print("\n1. Визначник det(A):", round(det_A, 4))
print("\n2. Обернена матриця A^(-1):\n", A_inv)
print("\n3. Перевірка (A @ A_inv):\n", check_product)
````

## Завдання 9. Розв'язання системи лінійних рівнянь

## Завдання 13. Функція з умовою

Створіть функцію

\[
f(x) =
\begin{cases}
x^2, & x < 0 \\
2x + 1, & x \geq 0
\end{cases}
\]

Застосуйте її до вектора значень від `-5` до `5`.

Спробуйте реалізувати функцію двома способами:
1. через звичайну функцію Python;
2. через `np.where()`.

````python
x = np.arange(-5, 6)

# Спосіб 1
def f_where(x_arr):
    return np.where(x_arr < 0, x_arr**2, 2 * x_arr + 1)

# Спосіб 2
def f_python_scalar(val):
    if val < 0:
        return val**2
    else:
        return 2 * val + 1

f_python = np.vectorize(f_python_scalar)

res_where = f_where(x)
res_python = f_python(x)

print("Вхідний вектор x:              ", x)
print("Результат (через np.where):   ", res_where)
print("Результат (через Python func): ", res_python)

````

## Завдання 15. Підсумкове

Згенеруйте матрицю `5 × 5` із випадкових цілих чисел від `1` до `100`.

Виконайте такі дії:
1. Знайдіть середнє значення кожного рядка.
2. Знайдіть середнє значення кожного стовпця.
3. Замініть усі елементи, менші за загальне середнє матриці, на `0`.
4. Знайдіть рядок із найбільшою сумою.
5. Нормалізуйте початкову матрицю за формулою:

\[
x_{norm} = \frac{x - x_{min}}{x_{max} - x_{min}}
\]

6. Напишіть функцію, яка виконує таку нормалізацію для довільної числової матриці.

````python
np.random.seed(42)

matrix = np.random.randint(1, 101, size=(5, 5))

row_means = np.mean(matrix, axis=1)

col_means = np.mean(matrix, axis=0)

overall_mean = np.mean(matrix)
matrix_with_zeros = matrix.copy()
matrix_with_zeros[matrix_with_zeros < overall_mean] = 0

row_sums = np.sum(matrix, axis=1)
max_row_index = np.argmax(row_sums)
max_row = matrix[max_row_index]

def normalize_matrix(mat):
    """
    Нормалізує числову матрицю за формулою: (x - x_min) / (x_max - x_min)
    """
    mat = np.array(mat, dtype=float)
    x_min = np.min(mat)
    x_max = np.max(mat)
    return (mat - x_min) / (x_max - x_min)

matrix_normalized = normalize_matrix(matrix)

print("Початкова матриця 5x5:\n", matrix)
print("\n1. Середнє значення кожного рядка:", row_means)
print("2. Середнє значення кожного стовпця:", col_means)
print(f"   (Загальне середнє матриці: {overall_mean:.2f})")
print("\n3. Матриця після заміни елементів < середнього на 0:\n", matrix_with_zeros)
print(f"\n4. Рядок з найбільшою сумою (Індекс {max_row_index}, сума = {row_sums[max_row_index]}):", max_row)
print("\n5. Нормалізована матриця (значення від 0 до 1):\n", np.round(matrix_normalized, 4))

````
